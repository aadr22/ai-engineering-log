```python
import numpy as np
x = np.random.rand(1_000_000)

def sum_sq_loop(x):
    total = 0.0
    for v in x:
        total += v * v
    return total
```

```python
def sum_sq_vec(x):
    return np.sum(x * x)    # square every element at once, then add them all

print(np.allclose(sum_sq_loop(x), sum_sq_vec(x)))
%timeit sum_sq_loop(x)
%timeit sum_sq_vec(x)
```

```python
X = np.random.rand(1000, 5) * np.array([1, 10, 100, 1000, 10000])

def zscore_loop(X):
    out = np.empty_like(X)
    print(X.shape[1])
    for j in range(X.shape[1]):
        col = X[:, j]
        mean = sum(col) / len(col)
        print(mean.shape)
        std = (sum((c - mean) ** 2 for c in col) / len(col)) ** 0.5
        for i in range(X.shape[0]):
            out[i, j] = (X[i, j] - mean) / std
    return out
```

```python
def zscore_vec(X):
    return (X - X.mean(axis=0)) / X.std(axis=0)

print(np.allclose(zscore_loop(X), zscore_vec(X)))
%timeit zscore_loop(X)
%timeit zscore_vec(X)
```

```python
a = np.random.randn(1_000_000)
def clip_neg_loop(a):
    out = a.copy()
    for i in range(len(out)):
        if out[i] < 0:
            out[i] = 0
    return out
```

```python
def clip_neg_mask(a):
    out = a.copy()          # copy so the original stays untouched
    out[out < 0] = 0        # mask picks the negatives, assignment zeroes them
    return out

def clip_neg_max(a):
    return np.maximum(a, 0) # element-wise larger of each value and 0: this is ReLU

print(np.allclose(clip_neg_loop(a), clip_neg_mask(a)))
print(np.allclose(clip_neg_loop(a), clip_neg_max(a)))
%timeit clip_neg_loop(a)
%timeit clip_neg_mask(a)    
%timeit clip_neg_max(a)    
```

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])       # (2, 3)
B = np.array([[1, 0],
              [0, 1],
              [1, 1]])          # (3, 2)
```

```python
print(A @ B)               
```

```python
print(A * B)
```

```python
print(A @ A.T)              # (2, 3) @ (3, 2) -> (2, 2)
```

```python
print(A.T @ A)              # (3, 2) @ (2, 3) -> (3, 3)
```

```python
rng = np.random.default_rng(seed=7)
scores = rng.integers(40, 100, size=(4, 6))
tests = np.array(["math", "physics", "chem", "bio", "english", "history"])
```

```python
best_index = np.argmax(scores, axis=1)
```

```python
best_tests = tests[best_index]
```

```python
print("Scores:\n", scores)
print("Best test indices:", best_index)
print("Best test names:", best_tests)
```

```python
order = np.argsort(scores[0])
student_0_tests = tests[order]
```

```python
print("Student 0, worst → best:", student_0_tests)
```
